```python
# Calculate weekly revenue and transaction count
weekly_revenue = (
    df.dropna(subset=['Transaction Date', 'Total Spent'])
      .groupby('Week')
      .agg(
          Revenue=('Total Spent', 'sum'),
          Transactions=('Transaction ID', 'count')
      )
      .reset_index()
)

weekly_revenue.head()
```

```python
#to calculate avg and std dev weekly revenue
mean_revenue = weekly_revenue['Revenue']
std_revenue = weekly_revenue['Revenue']
```

```python
# Calculate each week's z-score
weekly_revenue['Z_score'] = (
    weekly_revenue['Revenue'] - mean_revenue
) / std_revenue
```

```python
# Flag anomalies
weekly_revenue['Anomaly'] = weekly_revenue['Z_score'].abs() > 2

# Display anomaly weeks
weekly_revenue[weekly_revenue['Anomaly']]
```

```python
# Calculate monthly revenue
monthly_revenue = (
    df.dropna(subset=['Transaction Date', 'Total Spent'])
      .groupby(df['Transaction Date'].dt.to_period('M'))['Total Spent']
      .sum()
      .reset_index()
)

monthly_revenue.columns = ['Month', 'Revenue']

# Calculate the previous month's revenue
monthly_revenue['Previous Revenue'] = monthly_revenue['Revenue'].shift(1)

# Identify whether revenue fell
monthly_revenue['Declined'] = (
    monthly_revenue['Revenue'] < monthly_revenue['Previous Revenue']
)

monthly_revenue
```

```python
# Count consecutive months of declining revenue
monthly_revenue['Decline_Group'] = (
    monthly_revenue['Declined'] != monthly_revenue['Declined'].shift()
).cumsum()

# Count consecutive declining months
monthly_revenue['Consecutive_Declines'] = (
    monthly_revenue['Declined']
    .groupby(monthly_revenue['Decline_Group'])
    .cumsum()
)

# Show months where revenue declined
monthly_revenue[monthly_revenue['Declined']]
```

```python
### Prediction Model Limitation

The loan dataset contains 20,000 records, but all borrowers have a Default_Status of 1 (100%). There is no variation in the target variable.

Consequently, Logistic Regression, KNN, ROC-AUC, cumulative gains, and top 20% default capture cannot be meaningfully evaluated. SMOTE or class weighting cannot resolve the absence of a second outcome class.

The analysis therefore focuses on exploratory borrower risk indicators rather than developing a predictive default model.
```

```python
### Early Warning Indicators
The following indicators are relevant for monitoring potential borrower risk:

1. **Payment Delays:** Frequent repayment delays may signal emerging repayment difficulties.
2. **Credit Score:** Low credit scores may indicate weaker creditworthiness.
3. **Debt-to-Income Ratio:** High DTI may indicate that a borrower has a heavy debt burden relative to income.
4. **Existing Debt:** High outstanding debt may reduce a borrower's repayment capacity.
5. **Credit Utilisation:** High credit utilisation may indicate increased reliance on available credit.

These are exploratory indicators and cannot be confirmed as predictors of default using this dataset because all records have the same default outcome.
    
```
